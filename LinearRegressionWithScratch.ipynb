**Linear Regression With scratch**

```python
import numpy as np

# Function to compute the cost (Mean Squared Error)
def compute_cost(X, y, theta):
    m = len(y)
    y_pred = X.dot(theta)
    cost = (1/2*m) * np.sum(np.square(y_pred - y))
    return cost

# Function to perform gradient descent to learn the parameters theta
def gradient_descent(X, y, theta, learning_rate, iterations):
    m = len(y)
    cost_history = np.zeros(iterations)

    for i in range(iterations):
        # Predicted value
        y_pred = X.dot(theta)

        # Gradient calculation
        gradient = (1/m) * X.T.dot(y_pred - y)

        # Updating the parameters
        theta = theta - learning_rate * gradient

        # Store the cost for each iteration
        cost_history[i] = compute_cost(X, y, theta)

    return theta, cost_history

# Generating some random data for demonstration
np.random.seed(0)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

# Adding a column of ones to X (for the intercept term)
X_b = np.c_[np.ones((100, 1)), X]  # X_b is the new X with intercept term
theta = np.random.randn(2, 1)  # random initialization of theta

# Setting hyperparameters for gradient descent
learning_rate = 0.01
iterations = 1000

# Performing gradient descent
theta, cost_history = gradient_descent(X_b, y, theta, learning_rate, iterations)

print(f'Intercept: {theta[0]}')
print(f'Coefficient: {theta[1]}')

# Predicting the y values using the learned parameters
y_pred = X_b.dot(theta)

# Computing the cost with the learned parameters
cost = compute_cost(X_b, y, theta)
print(f'Final Cost: {cost}')
```
