**LogisticRegression With Scratch Implementation**

```python
import numpy as np

# Step 1: Load the dataset (we will create a synthetic dataset manually)
# Creating a synthetic dataset with 1000 samples, 10 features, and binary classification
np.random.seed(42)

# Generating random features (X) and random binary labels (y)
X = np.random.randn(1000, 10)  # 1000 samples, 10 features
y = (np.random.rand(1000) > 0.5).astype(int)  # 1000 binary labels

# Step 2: Split the dataset into training and testing sets
def train_test_split(X, y, test_size=0.2):
    """Split the dataset into training and testing sets."""
    total_samples = X.shape[0]
    test_samples = int(test_size * total_samples)
    indices = np.random.permutation(total_samples)
    test_indices = indices[:test_samples]
    train_indices = indices[test_samples:]

    return X[train_indices], X[test_indices], y[train_indices], y[test_indices]

X_train, X_test, y_train, y_test = train_test_split(X, y)

# Step 3: Implement Logistic Regression with Gradient Descent
def sigmoid(z):
    """Sigmoid activation function."""
    return 1 / (1 + np.exp(-z))

def compute_cost(X, y, weights):
    """Compute the cost function for logistic regression."""
    m = len(y)
    h = sigmoid(np.dot(X, weights))
    cost = -(1 / m) * np.sum(y * np.log(h) + (1 - y) * np.log(1 - h))
    return cost

def gradient_descent(X, y, weights, learning_rate, num_iterations):
    """Perform gradient descent to optimize weights."""
    m = len(y)
    for i in range(num_iterations):
        h = sigmoid(np.dot(X, weights))
        gradient = np.dot(X.T, (h - y)) / m
        weights -= learning_rate * gradient

        # Optional: Print cost every 100 iterations
        if i % 100 == 0:
            cost = compute_cost(X, y, weights)
            print(f"Iteration {i}, Cost: {cost:.4f}")

    return weights

# Add an intercept term (bias) to X_train and X_test
X_train_with_intercept = np.hstack((np.ones((X_train.shape[0], 1)), X_train))
X_test_with_intercept = np.hstack((np.ones((X_test.shape[0], 1)), X_test))

# Initialize weights
weights = np.zeros(X_train_with_intercept.shape[1])

# Set hyperparameters
learning_rate = 0.01
num_iterations = 1000

# Train the model using gradient descent
weights = gradient_descent(X_train_with_intercept, y_train, weights, learning_rate, num_iterations)

# Step 4: Make predictions on the test data
def predict(X, weights):
    """Predict binary class labels based on a threshold of 0.5."""
    probabilities = sigmoid(np.dot(X, weights))
    return [1 if prob >= 0.5 else 0 for prob in probabilities]

y_pred = predict(X_test_with_intercept, weights)





# Step 6: Interpret the coefficients (optional)
# Coefficients represent the log-odds of the features
print("Feature Coefficients:")
for i, coef in enumerate(weights[1:]):
    print(f"Feature {i + 1}: {coef:.4f}")
```
