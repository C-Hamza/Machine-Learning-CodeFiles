**K Medoid Clustering**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn_extra.cluster import KMedoids
from sklearn.metrics import silhouette_score

# Generating random data for demonstration
np.random.seed(42)
n_samples = 300
n_features = 2
n_clusters = 4  # Number of clusters

X, y = make_blobs(n_samples=n_samples, n_features=n_features, centers=n_clusters, cluster_std=1.0)

# Creating the K-Medoids model
kmedoids = KMedoids(n_clusters=n_clusters, random_state=42)

# Fitting the model to the data
kmedoids.fit(X)

# Predicting the cluster labels
y_pred = kmedoids.predict(X)

# Getting the cluster centers (medoids)
medoids = kmedoids.cluster_centers_

# Calculating Silhouette Score
silhouette_avg = silhouette_score(X, y_pred)
print(f'Silhouette Score: {silhouette_avg}')

# Calculating Inertia (Sum of dissimilarities within clusters)
inertia = kmedoids.inertia_
print(f'Inertia: {inertia}')

# Plotting the clusters and cluster centers
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis', marker='o')
plt.scatter(medoids[:, 0], medoids[:, 1], c='red', marker='x')
plt.title('K-Medoids Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
